# CompareEdge SQL Analysis

## Introduction
CompareEdge is a SaaS product comparison website that lists and compares
software tools across multiple categories. This analysis explores the
comparedge.com database to identify the most valuable product categories,
highest priced products, currency distribution and feature richness across
331 products and 5 interconnected tables.

## Dataset
- Source: comparedge_v1.sqlite (SQLite database)
- Tables: categories, products, pricing_plans, features, price_history
- Total categories: 29 rows
- Total Products: 331 rows
- Total Pricing Plans: 1,013 rows
- Total features: 6052 rows
- Total price_history: 916 rows

## Skills Used
- SQL (SQLite)
- Python (Pandas)
- Multi-table JOIN queries

```python
#import libraries
import sqlite3
import pandas as pd

#connect to the database
conn = sqlite3.connect("comparedge_v1.sqlite")

#look what tables are inside
query = "SELECT name FROM sqlite_master WHERE type = 'table'"
tables = pd.read_sql(query, conn)
print(tables)

for table in ["categories", "products", "pricing_plans", "features", "price_history"]:
    query = f"SELECT COUNT(*) as rows FROM {table}"
    results = pd.read_sql(query, conn)
    print (f"{table}: {results['rows'][0]} rows")

#look whats inside each tables
for table in ["categories", "products", "pricing_plans", "features", "price_history"]:
    print(f"\n  {table}   ")
    query = f"SELECT * FROM {table} LIMIT 4"
    results = pd.read_sql(query, conn)
    print(results)

#Analysis
#Getting products names with their similar categories name.
query = "SELECT products.name, categories.name FROM products JOIN categories ON products.category_id = categories.id"
results = pd.read_sql(query, conn)
print(results)

#QN1.Which category with most products.
query = "SELECT categories.name, product_count FROM categories ORDER BY product_count DESC LIMIT 5"
results = pd.read_sql(query, conn)
print(results)

#QN2.Most expensive product monthly.
query = "SELECT products.name , price_monthly FROM pricing_plans JOIN products" \
" ON pricing_plans.product_id= products.id ORDER BY price_monthly DESC LIMIT 5"
results = pd.read_sql(query, conn)
print(results)

#QN3. Which currency appear mostly.
query = "SELECT currency,  COUNT(*) AS most_used FROM pricing_plans GROUP BY currency ORDER BY most_used DESC "
results = pd.read_sql(query, conn)
print(results)

#QN4.Which product has most features
query = """ SELECT products.name, COUNT(*) as features_counts FROM features JOIN products ON features.product_id
= products.id GROUP BY products.name ORDER BY features_counts DESC LIMIT 5"""
results = pd.read_sql(query, conn)
print(results)

#QN5.Category with highest average monthly price.
query = """SELECT categories.name, AVG(pricing_plans.price_monthly) AS AVRG_Price
FROM pricing_plans JOIN products ON pricing_plans.product_id = products.id
JOIN categories ON products.category_id = categories.id
GROUP BY categories.name ORDER BY AVRG_price DESC LIMIT 5 """
results = pd.read_sql(query, conn)
print(results)

#QN6.Product with biggest price range.
query = """SELECT products.name, MAX(price_history.price) - MIN(price_history.price) AS price_range
FROM price_history JOIN products ON price_history.product_id = products.id GROUP BY products.name ORDER BY price_range
 DESC LIMIT 5"""
results = pd.read_sql(query, conn)
print(results)
```
